```python
import pandas as pd
```

```python
air_quality_no2 = pd.read_csv("D:/pandas/air_quality_no2_long.csv",parse_dates=True)
air_quality_no2 = air_quality_no2[["date.utc","location","parameter","value"]]
air_quality_no2.head()
```

```python
air_quality_pm25 = pd.read_csv("D:/pandas/air_quality_pm25_long.csv",parse_dates=True)
air_quality_pm25 = air_quality_pm25[["date.utc","location","parameter","value"]]
air_quality_pm25.head()
```

```python
air_quality = pd.concat([air_quality_pm25,air_quality_no2],axis=0)
air_quality.head()
```

```python
print("Shape of the original pm25: ",air_quality_pm25.shape)
print("Shape of the original no2: ",air_quality_no2.shape)
print("Shape of the concatenated table: ",air_quality.shape)
```

```python
air_quality = air_quality.sort_values("date.utc")
air_quality.head()
```

```python
air_quality_ = pd.concat([air_quality_pm25,air_quality_no2], keys=["PM25","NO2"])
air_quality_
```

```python
air_quality.reset_index(level=0)
```
