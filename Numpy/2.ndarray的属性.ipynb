# ndarray的属性

```python
import numpy as np
```

```python
arr = np.array(1)
print(arr)
print("数组的形状：", arr.shape)
print("数组的维度：", arr.ndim)
print("元素的个数：", arr.size)
print("元素的数据类型：", arr.dtype)
```

```python
arr = np.array([1,2,3])
print(arr)
print("数组的形状：", arr.shape)
print("数组的维度：", arr.ndim)
print("元素的个数：", arr.size)
print("元素的数据类型：", arr.dtype)
```

```python
arr = np.array([[1,2.5,3], [4,5,6]])
print(arr)
print("数组的形状：", arr.shape)
print("数组的维度：", arr.ndim)
print("元素的个数：", arr.size)
print("元素的数据类型：", arr.dtype)
```
