# ndarray的创建

```python
import numpy as np
```

## 1.基础创建方法

```python
arr1 = np.array([1,2,3])
list = [[1,2,3], [4,5,6]]
arr2 = np.array(list, dtype=np.float64)
print(arr1)
print(arr2)
```

## 2.copy

```python
arr_copy = np.copy(arr1) # arr_copy is a copy of arr1, changes to arr_copy will not affect arr1
print(arr_copy)
arr_copy[0] = 10
print(arr1) # arr1 remains unchanged
print(arr_copy) # arr_copy has changed
```

## 3.预定义形状

```python
# 全0数组（默认是float64）
arr_zeros = np.zeros((2, 3)) # 2 rows, 3 columns
print(arr_zeros)
```

```python
# 全1数组（默认是float64）
arr_ones = np.ones((3, 2)) # 3 rows, 2 columns
print(arr_ones)
print(arr_ones.dtype) # 查看数据类型
```

```python
# 未初始化
arr_empty = np.empty((2, 3, 2)) # 2 rows, 3 columns, 2 depth
print(arr_empty)
```

```python
# 全为指定元素
arr_full = np.full((2, 3), 5) # 2 rows, 3 columns, all elements are 5
print(arr_full)
```

```python
# 和指定数组shape相同的全0/全1/全指定元素/未初始化数组
arr_zeros_like = np.zeros_like(arr1) # 与arr1形状相同的全0数组
arr_ones_like = np.ones_like(arr1)  # 与arr1形状相同的全1数组
arr_full_like = np.full_like(arr1, 5)  # 与arr1形状相同的全5数组
arr_empty_like = np.empty_like(arr1)  # 与arr1形状相同的未初始化数组
print(arr_zeros_like)
print(arr_ones_like)
print(arr_full_like)
print(arr_empty_like)
```

## 4.等差数列

```python
arr1 = np.arange(10) # 生成0到9的数组
arr2 = np.arange(1, 10, 2) # 生成1到9的奇数数组
arr3 = np.arange(10, 0, -1) # 生成10到1的数组
# 注意：arange的步长可以是负数，表示从大到小生成数组
# 包括起始值，不包括终止值
print(arr1)
print(arr2)
print(arr3)
```

## 5.等间隔数列

```python
arr= np.linspace(0, 1, 5) # 生成0到1之间的5个等间隔的数
# 和arange不同的是，linspace的终止值是包括在内的
print(arr)
```

## 6.等比数列

```python
arr_log = np.logspace(0, 2, 5) # 生成10^0到10^2之间的5个等比数
# 和linspace不同的是，logspace生成的是等比数列
print(arr_log)
# 为什么是10的0次方到10的2次方？因为logspace的底数是10，默认情况下，logspace生成的是10的幂次方
# 可以通过设置base参数来改变底数，例如：np.logspace(0, 2, 5, base=2)生成的是2^0到2^2之间的5个等比数
arr_log = np.logspace(1, 4, 4, base=2) # 生成2^1到2^4之间的4个等比数
# 实际上就相当于先算linspace(1,4,4)得到[1, 2, 3, 4]，然后再算2的这些次方
print(arr_log)
```

## 7.特殊矩阵

```python
# 单位矩阵
arr = np.eye(3, dtype=int) # 生成3x3的单位矩阵
print(arr)
```

```python
# 对角矩阵
arr = np.diag([1, 2, 3]) # 生成对角矩阵，主对角线为[1, 2, 3]
print(arr)
```

```python
# 随机数组的生成
# 生成0到1之间的随机数，形状为(2, 3)
arr_random = np.random.rand(2, 3)
print(arr_random)
```

```python
# 生成指定范围（3到6）内的随机浮点数，形状为(2, 3)
arr_random = np.random.uniform(3, 6, size=(2, 3))
print(arr_random)
```

```python
# 生成指定范围内的随机整数，形状为(2, 3)
arr_random_int = np.random.randint(1, 10, size=(2, 3)) # 生成1到9之间的随机整数，形状为(2, 3)
print(arr_random_int)
```

```python
# 生成随机数列（标准正态分布）
arr = np.random.randn(2, 3)
print(arr)
```

```python
# 设置随机数种子
# 有什么作用？设置随机数种子可以保证每次生成的随机数都是相同的，方便调试和复现结果。
# 为什么是42？随机数种子可以是任意整数，选择42只是为了示例，实际使用中可以根据需要选择不同的整数。
# 设为30作用是一样的，只是生成的随机数不同而已。
np.random.seed(42) # 设置随机数种子为42
arr = np.random.randint(1, 10, size=(2, 5)) # 生成1到9之间的随机整数，形状为(2, 5)
print(arr)
```
