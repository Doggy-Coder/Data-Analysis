# ndarray的特性

## 1.多维性

```python
import numpy as np
```

```python
arr1 = np.array(5) # 创建0维ndarray数组
print(arr1)
arr1.ndim # arr数组的维度
```

```python
arr2 = np.array([1,2]) # 创建1维ndarray数组
print(arr2)
arr2.ndim
```

```python
arr3 = np.array([[1,2,3], [4,5,6]]) # 创建2维ndarray数组
print(arr3)
arr3.ndim
```

## 2.同质性

### 由于同质性，数字1会变成字符1，整数1会变成小数1

```python
arr = np.array([1, 'hello']) # 不同的数据类型会被强制转换成相同的数据类型
print(arr)
```

```python
arr = np.array([1, 2.5]) # 不同的数据类型会被强制转换成相同的数据类型
print(arr)
```
